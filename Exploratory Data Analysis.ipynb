```python
import matplotlib.pyplot as plt

churn_counts = df["Churn"].value_counts()

plt.figure(figsize=(6,4))
plt.bar(churn_counts.index, churn_counts.values)
plt.xlabel("Customer Churn")
plt.ylabel("Number of Customers")
plt.title("Customer Churn Distribution")
plt.show()
```

```python
df["Contract"].value_counts()
```

```python
contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

contract_churn
```

```python
contract_churn["Yes"].plot(kind="bar", figsize=(8,5))

plt.xlabel("Contract Type")
plt.ylabel("Churn Rate (%)")
plt.title("Churn Rate by Contract Type")
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["tenure"], bins=30)

plt.xlabel("Tenure (Months)")
plt.ylabel("Number of Customers")
plt.title("Customer Tenure Distribution")

plt.show()
```

```python
df.groupby("Churn")["tenure"].mean()
```

```python
df.groupby("Churn")["MonthlyCharges"].mean()
```

```python
df.boxplot(column="MonthlyCharges", by="Churn", figsize=(8,5))

plt.title("Monthly Charges by Churn Status")
plt.suptitle("")
plt.xlabel("Churn")
plt.ylabel("Monthly Charges")

plt.show()
```

```python
internet_churn = pd.crosstab(
    df["InternetService"],
    df["Churn"],
    normalize="index"
) * 100

internet_churn
```

```python
internet_churn["Yes"].plot(
    kind="bar",
    figsize=(8,5)
)

plt.xlabel("Internet Service")
plt.ylabel("Churn Rate (%)")
plt.title("Churn Rate by Internet Service")
plt.xticks(rotation=0)

plt.show()
```

```python
churn_percentage = df["Churn"].value_counts(normalize=True) * 100

plt.figure(figsize=(7,5))
plt.bar(churn_percentage.index, churn_percentage.values)

plt.xlabel("Churn Status")
plt.ylabel("Percentage of Customers")
plt.title("Overall Customer Churn Rate")

for i, value in enumerate(churn_percentage.values):
    plt.text(i, value + 1, f"{value:.1f}%", ha="center")

plt.show()
```
